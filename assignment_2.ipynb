### Q1. Inspect the digits dataset and plot the images

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_digits
```

```python
digits = load_digits()
digits
```

```python
print(f"Dataset Shape: {digits['data'].shape}")
print(f"Image Shape: {digits['images'].shape}")
print(f"Image Labels: {digits['feature_names']}")
print(f"Columns in the dataset: {[x for x in digits.keys()]}")
```

```python
fig, ax = plt.subplots(figsize=(8,6))

sns.heatmap(digits['images'][0], cmap='bone', cbar=True, annot=True, ax=ax)
ax.set_title(f"Label: {digits['target'][0]}", fontsize=16, fontstyle='italic')

plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(18,8), sharex=True, sharey=True)

for i, ax in enumerate(axes.flat):
    sns.heatmap(digits['images'][i], cmap='bone', ax=ax, cbar=i == 9)
    ax.axis('off')
    ax.set_title(f"Label: {digits['target'][i]}", fontsize=16, fontstyle='italic')

plt.show()
```
